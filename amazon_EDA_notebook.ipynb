# Amazon Laptop Dataset - EDA Notebook
This notebook contains exploratory data analysis of the dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('amazon_laptops_dataset.csv')
df.head()
```

## Basic Info

```python
df.info()
```

## Descriptive Statistics

```python
df.describe()
```

## Visualizations

```python
brand_avg = df.groupby('Brand')['Price (₹)'].mean()
brand_avg.plot(kind='bar')
plt.title('Average Laptop Price by Brand')
plt.show()
```
